## Trinidad and Tobago Inflation Analysis (1991-2025)
This notebook analyses the Inflation of Trinidad and Tobago's economy over a 34 year period using the Consumer Price Index data sourced from the Central Bank of Trinidad and Tobago.

Inflation is a gradual loss of purchasing power that occurs due to the increase in the prices of goods and services over time. The Consumer Price Index (CPI) measures the average change in prices of a range of goods and services and is split into two parts, food inflation and core inflation, which behave differently in response to different economic forces.

Trinidad and Tobago's inflation mainly stems from two factors: its heavy dependence on food imports, which makes domestic prices vulnerable to global price shocks, and its managed exchange rate, which the Central Bank uses to influence import costs and broader price stability.

This notebook features three charts:
- Overall Inflation Rate (1991 - 2025)
- Food Inflation vs Core Inflation (1991 - 2025)
- The five highest years for inflation and what drove them

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = pd.read_csv('data/Consumer Price Index - Year-On-Year Per Cent Change (Annually).csv')
df.head()
df.columns.tolist()
df = df.rename(columns={
    'Date': 'date',
    'Year on Year Percent Change in the Consumer Price Index - Inflation Rate (%)': 'inflation',
    'Year on Year Percent Change in the Consumer Price Index - Food (%)': 'food_inflation',
    'Year on Year Percent Change in the Consumer Price Index - Core (%)': 'core_inflation'
})

df[['date', 'inflation', 'food_inflation', 'core_inflation']].head()
df['date'] = pd.to_datetime(df['date'], format='%d/%m/%Y')
date = df['date'].head()
```

## Trinidad & Tobago Inflation Rate (1991 - 2025)
Over the years, Trinidad and Tobago has experienced varied inflation rates, with this being tied to changes in global prices, global supply chain disruptions and domestic supply shocks. Between 1991 to 2025, Trinidad and Tobago experienced five key spikes in inflation. Of these spikes, the 1994 spike followed the 1993 currency devaluation, which drove up the cost of imports and filtered through to consumer prices over the following months. The remaining spikes; 2008, 2010, and 2012, were largely driven by surges in food and global commoditiy prices which affected Trinidad and Tobago's domestic prices due to its heavy dependence on food imports. Inflation spikes such as these erode purchasing power, which means that the same income buys less than it used to. This places a burden on lower income families who spend most of their earnings on food and basic goods.

```python

plt.figure(figsize=(12, 6))
plt.plot(df['date'], df['inflation'], color='blue', linewidth=2)



plt.gca().xaxis.set_major_locator(mdates.YearLocator(2)) 
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
plt.xticks(rotation=45)  

plt.title('Trinidad & Tobago Inflation Rate (1991-2025)')
plt.xlabel('Year')
plt.ylabel('Inflation Rate (%)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Trinidad and Tobago Food Inflation vs Core Inflation (1991 - 2025)
Food inflation measures price changes in food products whereas core inflation tracks the underlying price trends in the economy, disregarding the highly volatile food prices. Throughout the period, food inflation has vastly outpaced core inflation, which shows Trinidad and Tobago's high reliance on food imports. This reliance is highlighted through years such as 2008 where food inflation reached a peak of 25.9% and in other years such as 2010, and 2012. These surges in food prices all stem from global commodity shocks leaving households exposed to price increases driven by factors outside domestic policy control. This pattern across the period of 1991 - 2025 shows a structural vulnerability in Trinidad and Tobago's economy where global market volatility, rather than the domestic market, is the main driver of food price instability.

```python
plt.figure(figsize = (12,6))
plt.plot(df['date'], df['food_inflation'], color = 'red', label = 'Food Inflation', linewidth = 2)
plt.plot(df['date'], df['core_inflation'], color = 'blue', label = 'Core Inflation', linewidth = 2)

import matplotlib.dates as mdates

plt.gca().xaxis.set_major_locator(mdates.YearLocator(2)) 
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
plt.xticks(rotation=45)  

plt.title('Trinidad and Tobago Food Inflation vs Core Inflation (1991 - 2025)')
plt.xlabel('Year')
plt.ylabel('Inflation %')
plt.grid(True)
plt.legend(["Food Inflation", "Core Inflation"], loc = "upper right")
plt.tight_layout()
plt.show()
```

## T&T Top 5 Highest Inflation Years

**2008** - driven by a surge in global food prices which led to a peak food inflation of 25.9%

**1993** - currency devaluation driven by the government's decision to float the Trinidad and Tobago dollar (TTD) resulted in increased import costs.

**2010** - domestic supply shocks triggered by extreme weather conditions resulted in food inflation reaching peaks of 22.1% while core inflation remained at 4.3%, which confirms the crisis was isolated to the food supply chain.

**2012** - similar to 2010, weather disruptions caused severe domestic supply shortages which increased the price of locally produced goods. Additionally, higher food import costs contributed, resulting in a food inflation of 19.1%.

**1994** - the lingering effects of the 1993 currency flotation and resulting devaluation contributed to an inflation rate of 8.8%.

```python
top_years = df.nlargest(5, 'inflation')[['date', 'inflation']]
top_years['year'] = top_years['date'].dt.year

plt.figure(figsize=(10, 6))
plt.bar(top_years['year'].astype(str), top_years['inflation'], color='steelblue')
plt.title('T&T Top 5 Highest Inflation Years')
plt.xlabel('Year')
plt.ylabel('Inflation Rate (%)')
plt.tight_layout()
plt.show()
```

## Conclusion

Overall, T&T's inflation is shown to be highly volatile due to its high reliance on food imports, with this volatility stemming from global shocks rather than domestic policy failures.  This reliance is shown in the large gap between food inflation and core inflation, with food inflation being consistently higher than core inflation over the past 34 years. Because of this, households, especially those with lower income, are vulnerable to price increases on essential goods, which leads to increased spending on necessities as they now have to spend more of their income. Finally, the data leaves us with one question. Is it possible for T&T to overcome its reliance on food imports? Or is the government simply unable to enact measures to lower this dependence?
